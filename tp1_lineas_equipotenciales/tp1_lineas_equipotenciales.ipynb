# 1. Objetivo

Medir experimentalmente las líneas equipotenciales —los lugares donde el potencial es constante—
y, a partir de ellas, las líneas de campo, para dos formas distintas de poner los electrodos
sumergidos en agua y conectados a una fuente de baja tensión.

Las dos geometrías:

1. **Rectangular (placas paralelas):** dos láminas enfrentadas. Esperamos un potencial que crece
   parejo con la distancia y un campo uniforme.
2. **Polar (láminas a 90°):** una de las láminas girada 90°, de modo que forman un ángulo recto.
   Esperamos un potencial que crece parejo con el ángulo y un campo no uniforme, que se debilita
   como $1/r$ al alejarse del vértice.

En cada caso comparamos las equipotenciales medidas con el modelo ideal y vemos qué tanto se
cumple. Las líneas de campo no las medimos ni las comparamos por separado: quedan determinadas
porque son siempre perpendiculares a las equipotenciales.

# 2. Introducción y marco teórico

Lo que queremos es mapear el potencial y el campo eléctrico en una región plana. Para eso usamos
una cubeta con un dedo de agua y dos electrodos conectados a una fuente de continua. El agua
permite que exista una pequeña corriente entre los electrodos y, una vez que el sistema se
estabiliza, podemos medir el potencial eléctrico en diferentes puntos con un voltímetro.

Buscando, para varios voltajes distintos, los puntos que dan ese mismo valor y uniéndolos,
armamos las líneas equipotenciales. Son esas equipotenciales las que al final comparamos con lo
que predice el modelo ideal de cada configuración. Las líneas de campo no las medimos ni las
comparamos aparte: al ser siempre perpendiculares a las equipotenciales, quedan trazadas a
partir de ellas.

## 2.1. Campo eléctrico

El campo eléctrico en un punto es la fuerza que sentiría ahí una carga de prueba positiva $q_0$,
dividida por esa carga:

$$\vec{E}=\frac{\vec{F}}{q_0}$$

Se mide en N/C, que es lo mismo que V/m. Es un vector: tiene módulo, dirección y sentido. Por
convención apunta hacia donde se movería una carga positiva, así que las líneas de campo van
siempre de las zonas de mayor potencial a las de menor.

## 2.2. Potencial eléctrico

El potencial $V$ en un punto es la energía potencial eléctrica por unidad de carga:

$$V=\frac{U}{q_0}$$

A diferencia del campo, el potencial es un escalar: un solo número por punto. La diferencia de
potencial entre dos puntos $A$ y $B$ está ligada al **trabajo** $W_{AB}$ que hace la fuerza
eléctrica al mover una carga $q$ de $A$ a $B$ — trabajo, en física, es básicamente "cuánta
energía transfiere una fuerza cuando mueve algo de un lado a otro":

$$\Delta V=\frac{\Delta U}{q}=-\frac{W_{AB}}{q}$$

## 2.3. Relación entre potencial y campo

Si nos movemos un pasito $d\vec{l}$, el potencial cambia

$$dV=-\vec{E}\cdot d\vec{l}$$

y de ahí sale la relación general

$$\vec{E}=-\nabla V$$

$\nabla V$ ("el gradiente de $V$") es un vector armado con cuánto cambia $V$ en cada dirección.
En 2D y en cartesianas, eso se escribe

$$E_x=-\frac{\partial V}{\partial x},\qquad E_y=-\frac{\partial V}{\partial y}$$

donde $\partial V/\partial x$ es "cuánto cambia $V$ si te movés solo en $x$, dejando $y$ fijo"
(y lo mismo para $\partial V/\partial y$, con $x$ fijo). El signo menos dice que el campo apunta
hacia donde el potencial **cae** más rápido, y su módulo es justamente esa caída máxima por
unidad de longitud.

Cuando el campo es **uniforme** (como entre las placas paralelas de 2.5), esta relación se
integra directo entre dos puntos separados un desplazamiento $\vec d$ y queda la versión que
vimos en la teórica:

$$\Delta V=-\vec{E}\cdot\vec{d}$$

## 2.4. Líneas equipotenciales y líneas de campo

Una línea equipotencial junta todos los puntos que tienen el mismo potencial, $V=\text{cte}$.
Como el potencial no cambia a lo largo de esa línea, si nos movemos un pasito $d\vec l$ sobre
ella, esa variación es cero: $dV=0$. Pero en 2.3 vimos que $dV=-\vec E\cdot d\vec l$ siempre —
entonces, sobre la equipotencial, tiene que valer

$$\vec{E}\cdot d\vec{l}=0$$

Un producto escalar da cero cuando los dos vectores son perpendiculares entre sí. Como esto vale
para cualquier $d\vec l$ que sigamos sobre la línea, **el campo eléctrico es siempre
perpendicular a las líneas equipotenciales**. Esta es la idea que más usamos en este TP: una vez
que tenemos las equipotenciales medidas, ya sabemos para dónde apunta el campo en cada punto —
es la dirección perpendicular a la equipotencial que pasa por ahí — sin necesidad de medirlo
aparte.

También nos sirve para ver dónde el campo es más intenso. Si dibujamos varias equipotenciales
separadas siempre el mismo voltaje ($\Delta V$ fijo), donde se ven más **juntas** el campo es
más **fuerte**, y donde se separan, más débil. Esto sale de $|\vec{E}|\approx|\Delta V|/\Delta n$
($\Delta n$ = distancia entre una equipotencial y la siguiente): con el mismo $\Delta V$ arriba,
si $\Delta n$ es chico el cociente da grande, y si $\Delta n$ es grande, da chico.

Un caso particular que vamos a usar todo el tiempo: **una lámina metálica (un conductor) es,
ella sola, una equipotencial completa.** Dentro de un metal hay cargas que se mueven libremente;
si hubiera alguna diferencia de potencial entre dos puntos de la lámina, esas cargas se
correrían hasta cancelarla. Entonces, en equilibrio, toda la lámina queda al mismo voltaje — el
que le conectamos con la fuente. Por eso en 2.5 y 2.6 podemos decir directamente "la lámina
negativa está a $V=0$" y "la positiva está a $V=\Delta V$", sin tener que medir punto por punto
dentro del metal.

## 2.5. Modelo ideal — configuración rectangular (placas paralelas)

Dos láminas paralelas separadas una distancia $d$, con una diferencia de potencial $\Delta V$
entre ellas, y sin mirar los efectos de borde. Cada lámina es una equipotencial completa (2.4).
Poniendo la negativa en $x=0$ como referencia ($V=0$), con el eje $x$ perpendicular a las
láminas, el potencial es lineal en $x$ — la misma idea
que vimos en la teórica para el caso angular (2.6), pero acá con la distancia en vez del ángulo:

$$V=f(x)=m\cdot x$$

En $x=d$ (la lámina positiva) tiene que valer $\Delta V$:

$$\Delta V=m\cdot d\quad\Longrightarrow\quad m=\frac{\Delta V}{d}$$

**En nuestro caso** ($\Delta V=10$ V, $d=20$ cm, lámina $+$ en $x=20$ cm, lámina $-$ en $x=0$):

$$m=\frac{10}{20}=0{,}5\ \text{V/cm}\quad\Longrightarrow\quad V(x)=0{,}5\,x\quad[\text{V, con }x\text{ en cm}]$$

El campo es el gradiente cambiado de signo:

$$\vec{E}=-\nabla V=-m\,\hat{x}=(-0{,}5,\;0)\ \text{V/cm}$$

que es exactamente $\Delta V=-\vec{E}\cdot\vec{d}$ (2.3) para un desplazamiento $\vec d=d\,\hat x$
entre las dos láminas, con $\vec E$ constante.

**¿Cómo sacamos cada recta?** Para la equipotencial de un voltaje $V_k$ despejamos $x$ de
$V(x)=V_k$:

$$\boxed{\;x_k=\dfrac{V_k}{m}=\dfrac{d}{\Delta V}\,V_k=2\,V_k\ \text{[cm]}\;}$$

Como $x_k$ no depende de $y$, cada equipotencial ideal es una **recta vertical** $x=x_k$. Estas
son las rectas que dibujamos en el modelo ideal y en la comparación de la sección 4.1. Las
**líneas de campo** ideales son las horizontales $y=\text{cte}$ (perpendiculares a las
equipotenciales), que van de la lámina $+$ ($x=20$) a la $-$ ($x=0$); el módulo del campo es el
mismo en todos lados.

| Potencial $V_k$ | Recta equipotencial ideal |
|---:|:---|
| 2 V | $x=4$ cm |
| 3 V | $x=6$ cm |
| 4 V | $x=8$ cm |
| 5 V | $x=10$ cm |
| 7 V | $x=14$ cm |
| 8 V | $x=16$ cm |

## 2.6. Modelo ideal — configuración polar (láminas a 90°)

Al girar una lámina 90°, las dos quedan formando un ángulo recto y (idealmente) se juntan en un
vértice. Acá conviene pensar en polares $(r,\theta)$ con el origen en ese vértice. Cada lámina es
un conductor, así que —como vimos en 2.4— es una equipotencial completa: la negativa en
$\theta=0$ (referencia, $V=0$) y la positiva en $\theta=\theta_0=90^\circ=\pi/2$.

*Ojo: esto es al revés de la convención habitual de trigonometría (ahí $\theta$ arranca en el
eje $x$ y crece en sentido antihorario). Acá arranca en el eje $y'$ y crece en sentido horario
hacia el eje $x'$, porque elegimos que $\theta=0$ sea la lámina de referencia.*

Lejos de los bordes, por simetría el potencial no depende de $r$ sino solo del ángulo, y es
lineal en él — mismo esquema que en 2.5, cambiando $x$ por $\theta$ y $d$ por $\theta_0$:

$$V=f(\theta)=m\cdot\theta$$

En $\theta=\theta_0$ tiene que valer $\Delta V$:

$$\Delta V=m\cdot\theta_0\quad\Longrightarrow\quad m=\frac{\Delta V}{\theta_0}$$

**En nuestro caso** ($\Delta V=10$ V, $\theta_0=\pi/2$):

$$m=\frac{10}{\pi/2}\approx 6{,}37\ \text{V/rad}\quad\Longrightarrow\quad V(\theta)=6{,}37\,\theta$$

Para ubicar cada punto medido $(x',y')$ en este sistema usamos la conversión habitual a
polares, tomando $\theta$ desde el eje $y'$ hacia el eje $x'$:

$$r=\sqrt{x'^2+y'^2},\qquad \theta=\arctan\!\left(\frac{x'}{y'}\right)$$

Para el campo usamos la misma regla de 2.4: $|\vec E|\approx|\Delta V|/\Delta n$, con $\Delta n$
la distancia perpendicular entre equipotenciales. Acá las equipotenciales son los rayos a ángulo
constante, así que moverse perpendicular a un rayo, a una distancia $r$ del vértice, es recorrer
un arco de longitud $\Delta n=r\cdot\Delta\theta$ (arco = radio $\times$ ángulo, en radianes). En
ese arco el potencial cambia $\Delta V=m\cdot\Delta\theta$ (porque $V=m\cdot\theta$). Entonces:

$$|\vec{E}|\approx\frac{|\Delta V|}{\Delta n}=\frac{m\cdot\Delta\theta}{r\cdot\Delta\theta}
\quad\Longrightarrow\quad
\boxed{\;|\vec{E}|=\dfrac{m}{r}=\dfrac{\Delta V}{\theta_0\,r}\propto \dfrac{1}{r}\;}$$

El $\Delta\theta$ se cancela: por eso el campo no depende del ángulo, solo de $r$. Y como es
perpendicular a los rayos (regla de 2.4), va como un arco de circunferencia alrededor del
vértice. O sea que el campo ya **no es uniforme**: cerca del vértice es fuerte (el mismo salto de
potencial se reparte en un arco corto) y se va debilitando a medida que te alejás (ese mismo
salto se reparte en un arco cada vez más largo).

**¿Cómo sacamos cada recta?** Despejamos $\theta$ de $V(\theta)=V_k$:

$$\boxed{\;\theta_k=\dfrac{V_k}{m}=\dfrac{\theta_0}{\Delta V}\,V_k=\dfrac{\pi/2}{10}\,V_k\;}$$

Cada equipotencial ideal es la **semirrecta** $\theta=\theta_k$ que sale del vértice. En el
sistema $(x',y')$ (vértice en el origen, $\theta$ medido desde el eje $y'$ = lámina $-$) esa
semirrecta es

$$y'=\dfrac{x'}{\tan\theta_k}\qquad(x'\ge 0),$$

o en paramétrica $x'=s\,\sin\theta_k,\ \ y'=s\,\cos\theta_k$ con $s\ge 0$. Estas son las que
dibujamos en el modelo ideal y en la comparación 4.2. Las **líneas de campo** ideales son arcos
de circunferencia $x'^2+y'^2=\text{cte}$ (perpendiculares a las semirrectas), que van de la
lámina $+$ a la $-$.

  | Potencial $V_k$ | Semirrecta equipotencial ideal |
  |---:|:---|
  | 2 V | $\theta=18^\circ$ |
  | 3 V | $\theta=27^\circ$ |
  | 4 V | $\theta=36^\circ$ |
  | 5 V | $\theta=45^\circ$ |
  | 7 V | $\theta=63^\circ$ |
  | 8 V | $\theta=72^\circ$ |

```python
import warnings
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import splprep, splev

warnings.filterwarnings("ignore")
plt.rcParams["figure.dpi"] = 110

# ---- Parametros comunes del experimento ----
dV = 10.0                 # diferencia de potencial aplicada [V]
d_placas = 20.0           # separacion entre laminas, config 1 [cm]
alto_bandeja = 20.0       # alto real del papel milimetrado, config 1 [cm] (bandeja de 20x20)
theta0 = np.pi / 2        # angulo entre laminas, config 2 [rad]
niveles_V = [2, 3, 4, 5, 7, 8]

# ---- Geometria config 1 (placas paralelas) ----
# Se dibujan como lineas completas (igual que en config 2), no como un punto: la lamina real
# es una tira vertical que ocupa todo el alto de la bandeja, no tiene una "altura" propia.
electrodo_pos = [(20, 0), (20, alto_bandeja)]   # lamina + : sobre x=20
electrodo_neg = [(0, 0), (0, alto_bandeja)]     # lamina - : sobre x=0 (referencia)

# ---- Geometria config 2 (laminas a 90 grados) ----
# Para dejar el grafico con el vertice en el ORIGEN (abajo-izquierda) se trabaja en el sistema
# con el vertice en (0, 0):  x' = 25 - x ,  y' = 15 - y   (reflexion respecto del vertice).
vertice_medido = (25.0, 15.0)        # interseccion estimada de las laminas en el papel
vertice = (0.0, 0.0)                 # vertice en el sistema del grafico
lamina_pos = [(0, 0), (20, 0)]       # lamina + : sobre el eje x' (grafico simplificado, ideal)
lamina_neg = [(0, 0), (0, 15)]       # lamina - : sobre el eje y' (grafico simplificado, ideal)


def a_grafico(pts):
    "Refleja los puntos medidos (x,y) al sistema (x',y') con el vertice en el origen."
    P = np.asarray(pts, dtype=float)
    return np.column_stack([vertice_medido[0] - P[:, 0], vertice_medido[1] - P[:, 1]])


def ordenar_por_eje_principal(pts):
    "Ordena una nube de puntos a lo largo de su direccion principal (PCA)."
    pts = np.asarray(pts, dtype=float)
    c = pts.mean(axis=0)
    _, _, vt = np.linalg.svd(pts - c)
    return pts[np.argsort((pts - c) @ vt[0])]


def curva_suave(pts, n=200, s=None):
    "Spline suave a traves de los puntos (ya ordenados). s>0 filtra el ruido de lectura."
    pts = np.asarray(pts, dtype=float)
    if len(pts) < 4:
        return pts[:, 0], pts[:, 1]
    s = 0.8 * len(pts) if s is None else s
    tck, _ = splprep([pts[:, 0], pts[:, 1]], s=s)
    return splev(np.linspace(0, 1, n), tck)


def a_polares(pts, v=None):
    "Convierte puntos (x',y') a (r, theta). theta=0 sobre la lamina - (eje y'), pi/2 sobre la + (eje x')."
    v = vertice if v is None else v
    pts = np.asarray(pts, dtype=float)
    dx, dy = pts[:, 0] - v[0], pts[:, 1] - v[1]
    return np.hypot(dx, dy), np.arctan2(dx, dy)
```

## 2.7. Gráficos del modelo ideal

Antes de meter las mediciones, dibujamos cómo **debería** verse cada configuración según la
teoría. El código evalúa las fórmulas que dedujimos arriba: $V(x)=\frac{\Delta V}{d}\,x$ para el
caso rectangular, y $V(\theta)=\frac{\Delta V}{\theta_0}\,\theta$ para el polar (usando
$\theta=\arctan(x'/y')$ para ubicar cada punto). Con esos valores de $V$, cada gráfico muestra
tres cosas:

- el **potencial**, como un mapa de color;
- las **equipotenciales**, en blanco — son los puntos donde $V$ vale lo mismo, es decir las
  rectas $x=2\,V_k$ (rectangular) y las semirrectas $\theta=\theta_k$ (polar) que dedujimos en
  2.5 y 2.6;
- el **campo eléctrico**, en gris.

```python
# --- Modelo ideal: configuracion rectangular (placas paralelas) ---
x = np.linspace(0, 20, 200)
y = np.linspace(0, alto_bandeja, 200)
X, Y = np.meshgrid(x, y)
V_ideal = (dV / d_placas) * X            # V(x) = 0.5 x   (no depende de y)

fig, ax = plt.subplots(figsize=(7, 6))
mapa = ax.contourf(X, Y, V_ideal, levels=20, cmap="viridis")
fig.colorbar(mapa, label="V ideal [V]")
ax.contour(X, Y, V_ideal, levels=niveles_V, colors="white", linewidths=1.2)
for lv in niveles_V:                      # etiqueta en la mitad de cada recta blanca
    ax.text(2 * lv, 9, f"{lv} V", fontsize=8, ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="0.6", lw=0.5))

Ex = np.full_like(X, -dV / d_placas)
Ey = np.zeros_like(Y)
ax.streamplot(X, Y, Ex, Ey, color="0.2", density=0.6, linewidth=0.8, arrowsize=1.2)

ep, en = np.array(electrodo_pos), np.array(electrodo_neg)
ax.plot(ep[:, 0], ep[:, 1], color="red", linewidth=4, solid_capstyle="round", label="Lámina A (+)")
ax.plot(en[:, 0], en[:, 1], color="blue", linewidth=4, solid_capstyle="round", label="Lámina B (−)")
ax.set(xlabel="x [cm]", ylabel="y [cm]", title="Modelo ideal — placas paralelas")
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=2)
ax.set_aspect("equal")
plt.tight_layout()
plt.savefig("graficos/modelo_ideal_rectangular.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
# --- Modelo ideal: configuracion polar (laminas a 90 grados) ---
# Sistema con el vertice en el origen: lamina + sobre el eje x, lamina - sobre el eje y.
X_MAX, Y_MAX = 20.0, 15.0
xp = np.linspace(0, X_MAX, 220)
yp = np.linspace(0, Y_MAX, 220)
Xp, Yp = np.meshgrid(xp, yp)

Theta = np.arctan2(Xp, Yp)                 # 0 sobre el eje y (lamina -), pi/2 sobre el eje x (lamina +)
Rp = np.hypot(Xp, Yp)
V_ideal = np.where(Rp > 2.5, (dV / theta0) * Theta, np.nan)

dVy, dVx = np.gradient(V_ideal, yp, xp)
Exp, Eyp = -dVx, -dVy

fig, ax = plt.subplots(figsize=(7, 6))
mapa = ax.contourf(Xp, Yp, V_ideal, levels=np.linspace(0, 10, 11), cmap="viridis")
fig.colorbar(mapa, label="V ideal [V]")
ax.contour(Xp, Yp, V_ideal, levels=niveles_V, colors="white", linewidths=1.2)
ax.streamplot(Xp, Yp, Exp, Eyp, color="0.15", density=0.7, linewidth=0.8,
              arrowsize=1.0, broken_streamlines=False)

# etiqueta de voltaje en la mitad de cada recta blanca
for lv in niveles_V:
    th = theta0 / dV * lv
    r_ex = min(X_MAX / np.sin(th), Y_MAX / np.cos(th))   # donde el rayo sale de la caja
    rm = 0.5 * (2.5 + r_ex)
    ax.text(rm * np.sin(th), rm * np.cos(th), f"{lv} V", fontsize=8, ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="0.6", lw=0.5))

lp, ln = np.array(lamina_pos), np.array(lamina_neg)
ax.plot(lp[:, 0], lp[:, 1], color="red", linewidth=4, solid_capstyle="round", label="Lámina A (+)")
ax.plot(ln[:, 0], ln[:, 1], color="blue", linewidth=4, solid_capstyle="round", label="Lámina B (−)")
ax.plot(*vertice, "k*", markersize=13, label="Vértice")
ax.set(xlabel="x' [cm]", ylabel="y' [cm]", title="Modelo ideal — láminas a 90°")
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=3)
ax.set_aspect("equal")
plt.tight_layout()
plt.savefig("graficos/modelo_ideal_polar.png", dpi=200, bbox_inches="tight")
plt.show()
```

## 3.2. Montaje y procedimiento

1. Pusimos ~1 cm de agua en la bandeja.
2. Conectamos una lámina al $+$ de la fuente y la otra al $-$, con $\Delta V\approx 10$ V. La
   lámina negativa la tomamos como referencia ($V=0$).
3. Enganchamos el borne común (COM) del multímetro a la lámina negativa y la otra punta a una
   sonda que fuimos moviendo por el agua. Buscamos los puntos donde marcaba un voltaje prefijado
   ($V=2,\,3,\,4,\,5,\,7,\,8$ V) y para cada uno anotamos 8 coordenadas $(x,y)$ del papel
   milimetrado.
4. Los puntos de igual voltaje son cada línea equipotencial.
5. Repetimos todo para la segunda configuración, girando una lámina 90°.
6. Las líneas de campo no las medimos directo: se deducen de las equipotenciales, porque
   $\vec{E}$ siempre es perpendicular a ellas.

### 3.3.3. Potencial en función de la posición

Si el modelo vale, cada equipotencial tiene que ser vertical y su posición media $\bar{x}$ tiene
que crecer parejo con $V$. Ajustamos $V=a\,\bar{x}+b$ por **cuadrados mínimos** — el método
estándar para encontrar la recta que mejor pasa cerca de todos los puntos (`np.polyfit` lo hace
solo) — y comparamos la pendiente $a$ con la ideal $dV/dx=0{,}5$ V/cm. Como $E_x=-dV/dx$, esa
pendiente ya nos da una estimación del campo ($E_x=-a$). Además, para cada línea calculamos el
**desvío medio** respecto de su recta teórica $x_\text{ideal}=2V$: $\langle|x_i-x_\text{ideal}|\rangle$.

```python
vs = np.array(sorted(equip_rect))
x_medios = np.array([np.mean([p[0] for p in equip_rect[v]]) for v in vs])
x_std = np.array([np.std([p[0] for p in equip_rect[v]]) for v in vs])

a, b = np.polyfit(x_medios, vs, 1)
print(f"Ajuste experimental:  V(x) = {a:.3f} x + {b:.3f}")
print(f"Pendiente experimental dV/dx = {a:.3f} V/cm   (ideal: 0.500 V/cm)")
print(f"Campo estimado |Ex|         = {abs(a):.3f} V/cm   (ideal: 0.500 V/cm)")
print(f"Desviacion respecto del ideal: {100*(a-0.5)/0.5:+.1f} %")

print("\n  V [V] | x_medido | x_ideal | desvio medio |x_i - x_ideal|")
for v, xm in zip(vs, x_medios):
    dev = np.mean([abs(p[0] - 2 * v) for p in equip_rect[v]])
    print(f"   {v:>3.0f}  |  {xm:5.2f}   |  {2*v:5.2f}  |      {dev:.2f} cm")
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.errorbar(x_medios, vs, xerr=x_std, fmt="o", capsize=3, label="Datos experimentales")
xx = np.linspace(0, 20, 100)
ax.plot(xx, a * xx + b, "-", label=f"Ajuste: V = {a:.3f} x + {b:.3f}")
ax.plot(xx, 0.5 * xx, "--", label="Ideal: V = 0.5 x")
ax.set(xlabel="x [cm]", ylabel="Potencial V [V]",
       title="Potencial en función de la posición — placas paralelas")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.savefig("graficos/V_vs_x_rectangular.png", dpi=200, bbox_inches="tight")
plt.show()
```

### 3.3.4. Qué vemos en la configuración 1

- Los puntos de cada voltaje quedan **casi alineados en vertical**: las equipotenciales salen
  paralelas entre sí y a las láminas. El potencial depende de $x$ y casi nada de $y$, tal cual
  dice el modelo.
- El ajuste $V(\bar{x})$ da bien lineal, con pendiente cerca de los 0,5 V/cm ideales (ver los

### 3.4.3. Potencial en función del ángulo

Pasamos cada punto a polares $(r,\theta)$ respecto del vértice y graficamos $V$ contra $\theta$.
Si el modelo polar vale, los puntos tienen que caer sobre una recta $V(\theta)=a\,\theta+b$ con
pendiente ideal $\Delta V/\theta_0=10/(\pi/2)\approx 6{,}37$ V/rad. Para cada línea calculamos
también el **desvío medio angular** respecto de su semirrecta teórica
$\theta_\text{ideal}=\frac{\pi/2}{10}V$: $\langle|\theta_i-\theta_\text{ideal}|\rangle$. Los
puntos muy pegados a las láminas se salen de la recta, porque ahí la equipotencial deja de ser
radial y se pone paralela a la superficie del conductor.

```python
def angulos_medios(equip):
    vs = np.array(sorted(equip))
    thm = np.array([np.degrees(np.mean(a_polares(equip[v])[1])) for v in vs])
    return vs, thm


def pendiente(equip):
    vs, thm = angulos_medios(equip)
    a, b = np.polyfit(np.radians(thm), vs, 1)
    return a, b


vs_p, th_medios = angulos_medios(equip_polar)
th_todos = np.concatenate([np.degrees(a_polares(equip_polar[v])[1]) for v in vs_p])
V_todos = np.concatenate([np.full(len(equip_polar[v]), v) for v in vs_p])

print("  V [V] | theta_medido | theta_ideal | desvio medio")
for v, tm in zip(vs_p, th_medios):
    th_id = np.degrees(theta0) / dV * v
    ths = np.degrees(a_polares(equip_polar[v])[1])
    dev = np.mean(np.abs(ths - th_id))
    print(f"   {v:>3.0f}  |   {tm:5.1f} deg  |  {th_id:5.1f} deg  |   {dev:4.1f} deg")

ap, bp = pendiente(equip_polar)
ideal = dV / theta0
print(f"\nAjuste: V(theta) = {ap:.2f} * theta + {bp:.2f}   (theta en rad)")
print(f"Pendiente experimental dV/dtheta = {ap:.2f} V/rad   (ideal: {ideal:.2f} V/rad)")
print(f"Desviacion respecto del ideal: {100*(ap-ideal)/ideal:+.1f} %")
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(th_todos, V_todos, "o", alpha=0.35, label="Todos los puntos medidos")
ax.plot(th_medios, vs_p, "s", color="C3", markersize=8, label="Ángulo medio por equipotencial")
tt = np.linspace(0, 90, 100)
ax.plot(tt, ap * np.radians(tt) + bp, "-", label=f"Ajuste: V = {ap:.2f}·θ + {bp:.2f}")
ax.plot(tt, (dV / theta0) * np.radians(tt), "--", label=f"Ideal: V = {dV/theta0:.2f}·θ")
ax.set(xlabel="Ángulo θ desde la lámina − [grados]", ylabel="Potencial V [V]",
       title="Potencial en función del ángulo — láminas a 90°")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.savefig("graficos/V_vs_theta_polar.png", dpi=200, bbox_inches="tight")
plt.show()
```

# 5. Conclusiones

Para las dos geometrías pudimos medir las líneas equipotenciales con el multímetro en el agua y,
a partir de ellas, deducir cómo tienen que ser las líneas de campo —perpendiculares a las
equipotenciales, apuntando de mayor a menor potencial— sin necesidad de medirlas por separado.

En **placas paralelas** el experimento da lo que esperábamos: equipotenciales rectas y paralelas
a las láminas, potencial lineal en $x$ ($V\approx a\,x$, con $a$ cerca de 0,5 V/cm) y campo
prácticamente uniforme y horizontal, de $+$ a $-$; se cumple el modelo ideal. Lo que no cierra
del todo tiene explicación: la pendiente $dV/dx$ da un poco menos que 0,5 V/cm y con ordenada al
origen distinta de cero, algo compatible con una caída de tensión en los contactos lámina–agua;
las equipotenciales ondulan un poco cerca de los bordes porque las láminas no son infinitas
(efecto de borde); y los puntos tienen algo de dispersión por el error al ubicar la sonda
($\pm 1$ mm), la lectura del multímetro y que la capa de agua no es perfectamente pareja.

En **láminas a 90°** también sale el modelo: equipotenciales radiales desde el vértice,
potencial lineal en $\theta$ ($V\approx a\,\theta$) y campo en arcos, más intenso cerca del
vértice como $1/r$. Acá aparece la misma caída de tensión en los contactos lámina–agua, y además
las equipotenciales se curvan cerca de las láminas para
ponerse paralelas a ellas —como tiene que pasar sobre un conductor—, así que el modelo radial
vale solo en la zona central. Además, el vértice no es un punto que se mida directo: las láminas
no llegan hasta ahí, así que hay que estimar a ojo dónde se cruzarían si se las prolongara, y esa
estimación tiene un error de $\pm 1$–$2$ cm. Encima, al armar el experimento la lámina positiva
no quedó justo en el extremo de la negativa, sino un
poco más adentro (a un 25% de su largo, aproximadamente). Los dos errores pegan justo donde más
importan: cerca del vértice, que es donde están las equipotenciales de menor voltaje — por eso
son las que se ven más distorsionadas. El gráfico de abajo muestra la geometría ideal (sin ese
corrimiento) para no complicar el modelo.

Con el mismo $\Delta V$, el campo sale completamente distinto según la geometría: uniforme entre
placas paralelas, y $\propto 1/r$ entre láminas a 90°. Esa es la moraleja del trabajo: el campo
no lo decide solo el voltaje que ponés, sino sobre todo **cómo están puestos los conductores**.
