## Introducción
Un capacitor es un dispositivo formado por dos placas o láminas conductoras separadas entre sí por un dieléctrico especialmente diseñado para almacenar una cantidad finita de energía. La unidad de la capacidad es el Faradio ${[F]}$. 

Un dieléctrico es un medio eléctricamente aislante que se opone al paso de la corriente eléctrica.

Si se conecta el capacitor a un generador se produce un pasaje de electrones hasta que las dos placas consiguen albergar el mismo número de cargas pero de _signo contrario_. Una vez cargado, la corriente que circula por el circuito es nula y aunque se desconecte el generador, el capacitor no se descarga. 

La forma más fácil de armar un capacitor es colocar dos placas paralelas enfrentadas entre sí separadas por una distancia que es pequeña en comparación con sus dimensiones. Cuando las placas están cargadas se genera un campo eléctrico ${E}$ uniforme entre medio de las placas. La magnitud de ese campo eléctrico se puede calcular utilizando la Ley de Gauss o el principio de superposición de campos eléctricos, entonces:  ${E = \frac{\sigma}{\epsilon_{0}}}$, donde ${\epsilon_{0}}$ es la **permitividad eléctrica en el vacío**, con un valor de ${8,85.10^{-12} \frac{F}{m}}$.

Si se hace extensiva esta igualdad a toda la placa del capacitor, el campo eléctrico dentro del mismo queda definido como: ${E = \frac{q}{\epsilon_{0}.A'}}$, donde _q_ es la carga neta de cada una de las placas y _A_ es el área de cada una.

Dado que el campo es uniforme y la distancia entre las placas es _d_, la diferencia de potencial entre ambas será: $${\Delta V_{ab}= E . d = \frac{q}{\epsilon_{0}.A}}$$


Entonces, la **capacidad de un capacitor plano en el _vacío_** queda definida como: $${C= \frac{q}{\Delta V_{ab}} = \frac{\epsilon_{0}.A}{d}}$$


La capacitancia varia si, en vez de vacío, hay otro **tipo de _material_ entre las placas**. Esta se va a ver afectada por las características de este componente particular por lo que su ecuación cambiará a: $${C= \frac{\epsilon_{0}.k.A}{d} \Longrightarrow C= \frac{\epsilon_{0}.(1+X_{e}).A}{d}}$$
Allí $X_{e}$ va a ser la **constante dieléctrica del material** entre las placas.


## Procedimiento experimental

Para el desarrollo de este experimento se construyeron capacitores de placas paralelas con **dos placas metálicas de 20cm x 20cm** y **placas de vidrio o acrílico de _distintos espesores_** que se colocaron en el medio de las placas metálicas para generar distintas distancias entre estas. 
Las placas metálicas se conectaron a un multimetro digital para obtener los distintos valores de capacidad dependiendo de la **separación entre las placas**.




Se realizaron tres configuraciones de capacitores paralelos. 
* La **_primera configuración_** estaba compuesta por dos placas metálicas separadas con placas de vidrio de distintos espesores
* La **_segunda configuración_** por dos placas metálicas separadas con placas de acrílico. 
* La **_tercera configuración_** por dos placas metálicas separadas con pequeños cuadrados de acrílico que permitian el paso de aire entre las placas y que se ubicaron en cada esquina de las mismas.

Para _todas las configuraciones_ se tomaron **5 medidas de capacidad**, se armó una tabla y se procedió con la realización de gráficos de dispersión para poder determinar la ${X_{e}}$


## Mediciones y resultados
Para la **_Primera configuración_** se utilizaron dos placas metálicas con separaciones de vidrios de distintos espesores.

Vidrio | d [m] | C [F]
------------ | ------------ | -------------
1 placa C   | 0,0059m| ${3,26.10^{-10}}$ F
2 placa B  | 0,0038m| ${4,3.10^{-10}}$ F
3 placa B+C | 0,0097m| ${2,3.10^{-10}}$ F
4 placa Cx2 | 0,0118m | ${2,02.10^{-10}}$ F
5 placa Bx2 | 0,0076m| ${2,47.10^{-10}}$ F

Con los datos obtenidos se realizó un gráfico con los valores de la capacidad _C [F]_ por los valores de $\frac{1}{d}$ (_d_ es la separación entre las placas metálicas), para así determinar la constante de susceptibilidad dieléctrica $X_{e}$

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Valores de la tabla
VIDRIO_B = 1/0.0038
VIDRIO_C = 1/0.0059
VIDRIO_BB = 1/0.0076
VIDRIO_BC = 1/0.0097
VIDRIO_CC = 1/0.0118

CAPACIDAD_B = 4.3e-10
CAPACIDAD_C = 3.26e-10
CAPACIDAD_BB = 2.47e-10
CAPACIDAD_BC = 2.3e-10
CAPACIDAD_CC = 2.02e-10

d_values = [VIDRIO_C, VIDRIO_B, VIDRIO_BC, VIDRIO_CC, VIDRIO_BB]  # Valores de 1/d [m]
C_values = [CAPACIDAD_C, CAPACIDAD_B, CAPACIDAD_BC, CAPACIDAD_CC, CAPACIDAD_BB]  # Valores de C [F]

# Ajuste lineal (recta) a los datos
coef = np.polyfit(d_values, C_values, 1)
poly1d_fn = np.poly1d(coef)  # Función polinómica de ajuste

# Crear el gráfico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='orange', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la línea de ajuste
plt.plot(d_values, poly1d_fn(d_values), '--', color='purple', label="Ajuste lineal")

# Configuraciones del gráfico
plt.title('Vidrio')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 5.0e-10)
plt.legend()
plt.grid(True)

# Formato científico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje y
ax.xaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje x

# Ajustar márgenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuación de la recta y R² fuera del gráfico
m, b = coef
plt.text(1.5, 4.5e-10, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='purple', ha='right')
R2 = np.corrcoef(d_values, C_values)[0, 1]**2
plt.text(1.5, 4.3e-10, f'$R^2$ = {R2:.2f}', fontsize=14, color='purple', ha='right')

# Mostrar el gráfico
plt.show()
```


Comparando la formula de capacitancia (${C = \frac{\epsilon_{0}.k.A}{d}}$) con una recta (${y = m.x + b}$), se determinó  que ${\epsilon_{0}.k.A}$ es igual a la pendiente de la recta: $${m = \epsilon_{0}.k.A = 1,30.10^{-12}}$$

Para determinar el error de la pendiente se utilizo la siguiente fórmula: $${\Delta_{m} = \sqrt{\frac{m^{2}}{(N -2)}.(\frac{1}{R^{2}} - 1)}}$$
En dicha ecuación el valor de _N_ se corresponde con el número de mediciones realizadas. El gráfico nos proporcionó  el valor de la **pendiente _m_** y el **valor del coeficiente de determinación ${R^{2} = 0,99}$**, se reemplazaron ambos en la ecuación para encontrar el **error de la pendiente**: $${\Delta_{m} = \sqrt{\frac{(1,30.10^{-12})^{2}}{(5 -2)}.(\frac{1}{(0,99)} - 1)} = 7,54.10^{-14} = 0,075.10^{-12}}$$

Se llegó a la conclusión de que **la pendiente _m_ + su error** es igual a  $\Longrightarrow$   ${m = (1,30.10^{-12} \pm 0,075.10^{-12})}$

Utilizando la fórmula de la pendiente de la recta para despejar la **constante dieléctrica _k_** se llegó a que $${m = \epsilon_{0}.k.A \Longrightarrow k= \frac{m}{A.\epsilon_{0}}}$$

Entonces, reemplazando por:
1. **Permitividad del vacío** $\epsilon_{0}$: ${8,85.10^{-12} \frac{F}{m}}$
2. **Area de las placas metálicas**: ${20cm x 20cm = 0,2m x 0,2m= 0,04m^{2}}$
3. **Pendiente**: ${1,30.10^{-12}}$
$${k= \frac{1,30.10^{-12}}{0,04m^{2}.8,85.10^{-12}}= 3,64406796}$$

Se calculó el **error de la constante dieléctrica _k_**: $${\Delta_{m}= \Delta_{k}.\epsilon_{0}.A \Longrightarrow \Delta_{k}= \frac{\Delta_{m}}{\epsilon_{0}.A} =  \frac{0,075.10^{-12}}{8,85.10^{-12}.0,04} = 0,21}$$

Lo que nos deja como saldo final que **la constante dieléctrica _k_** es igual a $\Longrightarrow$ ${k = (4 \pm 0,2)}$. 
Utilizando la formula ${k = 1 + X_{e}}$ para calcular la susceptibilidad dieléctrica nos queda como resultado que: $${X_{e} = 3} \Longrightarrow {X_{e}= (3,0 \pm 0,2)}$$

Con los datos obtenidos se realizó un gráfico con los valores de la capacidad _C [F]_ por los valores de $\frac{1}{d}$ (_d_ es la separación entre las placas metálicas), para así determinar la constante de susceptibilidad dieléctrica $X_{e}$

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Valores de la nueva tabla (Acrílico)
ACRILICO_A = 1/0.0019
ACRILICO_D = 1/0.0037
ACRILICO_E = 1/0.0045
ACRILICO_ED = 1/0.0082
ACRILICO_EDA = 1/0.0101

CAPACIDAD_A = 4.55e-10
CAPACIDAD_D = 2.93e-10
CAPACIDAD_E = 2.3e-10
CAPACIDAD_ED = 1.47e-10
CAPACIDAD_EDA = 1.24e-10

d_values = [ACRILICO_D, ACRILICO_A, ACRILICO_E, ACRILICO_ED, ACRILICO_EDA]  # Valores de d [m]
C_values = [CAPACIDAD_D, CAPACIDAD_A, CAPACIDAD_E, CAPACIDAD_ED, CAPACIDAD_EDA]  # Valores de C [F]

# Ajuste lineal (recta) a los datos
coef = np.polyfit(d_values, C_values, 1)
poly1d_fn = np.poly1d(coef)  # Función polinómica de ajuste

# Crear el gráfico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='yellow', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la línea de ajuste
plt.plot(d_values, poly1d_fn(d_values), '--', color='blue', label="Ajuste lineal")

# Configuraciones del gráfico
plt.title('Acrílico')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 5.0e-10)
plt.legend()
plt.grid(True)

# Formato científico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje y

# Ajustar márgenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuación de la recta y R² fuera del gráfico
m, b = coef
plt.text(1.5, 4.5e-10, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='blue', ha='right')
R2 = np.corrcoef(d_values, C_values)[0, 1]**2
plt.text(1.5, 4.3e-10, f'$R^2$ = {R2:.2f}', fontsize=14, color='blue', ha='right')

# Mostrar el gráfico
plt.show()
```

1. **Pendiente** = ${7,74.10^{-13}}$
2. **Valor del coeficiente de determinación** ${R^{2}}$ = ${0,99}$
3. **Numero de mediciones _N_** = ${5}$
4. **Permitividad del vacío** $\epsilon_{0}$ = ${8,85.10^{-12} \frac{F}{m}}$
5. **Area de las placas metálicas**: ${20cm x 20cm = 0,2m x 0,2m= 0,04m^{2}}$

Se calcula el **error de la pendiente $\Delta_{m}$**: $${\Delta_{m} = \sqrt{\frac{m^{2}}{(N -2)}.(\frac{1}{R^{2}} - 1)}}  \Longrightarrow \sqrt{\frac{(7,74.10^{-13})^{2}}{(5 -2)}.(\frac{1}{0,99} - 1)} = 4,49.10^{-14} = 0,45.10^{-13}$$

La **pendiente + su error** quedan definidos de la siguiente manera: ${m = (7,74.10^{-13} \pm 0,45.10^{-13})}$

Utilizando la fórmula de la pendiente de la recta para despejar la **constante dieléctrica _k_** se llegó a que $${m = \epsilon_{0}.k.A \Longrightarrow k= \frac{m}{A.\epsilon_{0}} = \frac{7,74.10^{-13}}{0,04m^{2}.8,85.10^{-12}} = 2,186440678}$$

Se calculó el **error de la constante dieléctrica _k_**: $${\Delta_{m}= \Delta_{k}.\epsilon_{0}.A \Longrightarrow \Delta_{k}= \frac{\Delta_{m}}{\epsilon_{0}.A} =  \frac{0,45.10^{-13}}{8,85.10^{-12}.0,04} = 0,1271186440}$$

Lo que nos deja como saldo final que **la constante dieléctrica _k_** es igual a $\Longrightarrow$ ${k = (2 \pm 0,1)}$. 
Utilizando la formula ${k = 1 + X_{e}}$ para calcular la susceptibilidad dieléctrica nos queda como resultado que: $${X_{e} = 1} \Longrightarrow {X_{e}= (1,0 \pm 0,1)}$$
