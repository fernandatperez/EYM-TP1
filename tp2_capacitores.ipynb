## Introducción
Un capacitor es un dispositivo formado por dos placas o láminas conductoras separadas entre sí por un dieléctrico especialmente diseñado para almacenar una cantidad finita de energía. La unidad de la capacidad es el Faradio ${[F]}$. 

Un dieléctrico es un medio eléctricamente aislante que se opone al paso de la corriente eléctrica.

Si se conecta el capacitor a un generador se produce un pasaje de electrones hasta que las dos placas consiguen albergar el mismo número de cargas pero de _signo contrario_. Una vez cargado, la corriente que circula por el circuito es nula y aunque se desconecte el generador, el capacitor no se descarga. 

La forma más fácil de armar un capacitor es colocar dos placas paralelas enfrentadas entre sí separadas por una distancia que es pequeña en comparación con sus dimensiones. Cuando las placas están cargadas se genera un campo eléctrico ${E}$ uniforme entre medio de las placas. La magnitud de ese campo eléctrico se puede calcular utilizando la Ley de Gauss o el principio de superposición de campos eléctricos, entonces:  ${E = \frac{\sigma}{\epsilon_{0}}}$, donde ${\epsilon_{0}}$ es la **permitividad eléctrica en el vacío**, con un valor de ${8,85.10^{-12} \frac{F}{m}}$.

Si se hace extensiva esta igualdad a toda la placa del capacitor, el campo eléctrico dentro del mismo queda definido como: ${E = \frac{q}{\epsilon_{0}.A}}$, donde _q_ es la carga neta de cada una de las placas y _A_ es el área de cada una.

Dado que el campo es uniforme y la distancia entre las placas es _d_, la diferencia de potencial entre ambas será: $${\Delta V_{ab}= E . d = \frac{q.d}{\epsilon_{0}.A}}$$


Entonces, la **capacidad de un capacitor plano en el _vacío_** queda definida como: $${C= \frac{q}{\Delta V_{ab}} = \frac{\epsilon_{0}.A}{d}}$$


La capacitancia varía si, en vez de vacío, hay otro **tipo de _material_ entre las placas**. Esta se va a ver afectada por las características de este componente particular por lo que su ecuación cambiará a: $${C= \frac{\epsilon_{0}.k.A}{d} \Longrightarrow C= \frac{\epsilon_{0}.(1+X_{e}).A}{d}}$$
Allí _k_ es la **constante dieléctrica del material** entre las placas y $X_{e}$ es su **susceptibilidad dieléctrica**, relacionadas por ${k = 1 + X_{e}}$.


## Procedimiento experimental

Para el desarrollo de este experimento se construyeron capacitores de placas paralelas con **dos placas metálicas de 20cm x 20cm** y **placas de vidrio o acrílico de _distintos espesores_** que se colocaron en el medio de las placas metálicas para generar distintas distancias entre estas. 
Las placas metálicas se conectaron a un multímetro digital para obtener los distintos valores de capacidad dependiendo de la **separación entre las placas**.




Se realizaron tres configuraciones de capacitores paralelos. 
* La **_primera configuración_** estaba compuesta por dos placas metálicas separadas con placas de vidrio de distintos espesores
* La **_segunda configuración_** por dos placas metálicas separadas con placas de acrílico. 
* La **_tercera configuración_** por dos placas metálicas separadas con pequeños cuadrados de acrílico que permitían el paso de aire entre las placas y que se ubicaron en cada esquina de las mismas.

Para _todas las configuraciones_ se tomaron **5 medidas de capacidad**, se armó una tabla y se procedió con la realización de gráficos de dispersión para poder determinar la ${X_{e}}$


## Mediciones y resultados
Para la **_Primera configuración_** se utilizaron dos placas metálicas con separaciones de vidrios de distintos espesores.

Vidrio | d [m] | C [F]
------------ | ------------ | -------------
1 placa C   | 0,0059m| ${3,26.10^{-10}}$ F
2 placa B  | 0,0038m| ${4,3.10^{-10}}$ F
3 placa B+C | 0,0097m| ${2,3.10^{-10}}$ F
4 placa Cx2 | 0,0118m | ${2,02.10^{-10}}$ F
5 placa Bx2 | 0,0076m| ${2,47.10^{-10}}$ F

Con los datos obtenidos se realizó un gráfico con los valores de la capacidad _C [F]_ por los valores de $\frac{1}{d}$ (_d_ es la separación entre las placas metálicas), para así determinar la susceptibilidad dieléctrica $X_{e}$

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Valores de la tabla
VIDRIO_B = 1/0.0038
VIDRIO_C = 1/0.0059
VIDRIO_BB = 1/0.0076
VIDRIO_BC = 1/0.0097
VIDRIO_CC = 1/0.0118

CAPACIDAD_B = 4.3e-10
CAPACIDAD_C = 3.26e-10
CAPACIDAD_BB = 2.47e-10
CAPACIDAD_BC = 2.3e-10
CAPACIDAD_CC = 2.02e-10

d_values = [VIDRIO_C, VIDRIO_B, VIDRIO_BC, VIDRIO_CC, VIDRIO_BB]  # Valores de 1/d [m]
C_values = [CAPACIDAD_C, CAPACIDAD_B, CAPACIDAD_BC, CAPACIDAD_CC, CAPACIDAD_BB]  # Valores de C [F]

# Ajuste lineal (recta) a los datos
coef = np.polyfit(d_values, C_values, 1)
poly1d_fn = np.poly1d(coef)  # Función polinómica de ajuste

# Crear el gráfico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='orange', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la línea de ajuste
plt.plot(d_values, poly1d_fn(d_values), '--', color='purple', label="Ajuste lineal")

# Configuraciones del gráfico
plt.title('Vidrio')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 5.0e-10)
plt.legend()
plt.grid(True)

# Formato científico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje y
ax.xaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje x

# Ajustar márgenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuación de la recta y R² fuera del gráfico
m, b = coef
plt.text(1.5, 4.5e-10, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='purple', ha='right')
R2 = np.corrcoef(d_values, C_values)[0, 1]**2
plt.text(1.5, 4.3e-10, f'$R^2$ = {R2:.2f}', fontsize=14, color='purple', ha='right')

# Mostrar el gráfico
plt.show()
```


Comparando la fórmula de capacitancia (${C = \frac{\epsilon_{0}.k.A}{d}}$) con una recta (${y = m.x + b}$), se determinó  que ${\epsilon_{0}.k.A}$ es igual a la pendiente de la recta: $${m = \epsilon_{0}.k.A = 1,30.10^{-12}}$$

Para determinar el error de la pendiente se utilizó la siguiente fórmula: $${\Delta_{m} = \sqrt{\frac{m^{2}}{(N -2)}.(\frac{1}{R^{2}} - 1)}}$$
En dicha ecuación el valor de _N_ se corresponde con el número de mediciones realizadas. El gráfico nos proporcionó  el valor de la **pendiente _m_** y el **valor del coeficiente de determinación ${R^{2} = 0,99}$**, se reemplazaron ambos en la ecuación para encontrar el **error de la pendiente**: $${\Delta_{m} = \sqrt{\frac{(1,30.10^{-12})^{2}}{(5 -2)}.(\frac{1}{(0,99)} - 1)} = 7,54.10^{-14} = 0,075.10^{-12}}$$

Se llegó a la conclusión de que **la pendiente _m_ + su error** es igual a  $\Longrightarrow$   ${m = (1,30.10^{-12} \pm 0,075.10^{-12})}$

Utilizando la fórmula de la pendiente de la recta para despejar la **constante dieléctrica _k_** se llegó a que $${m = \epsilon_{0}.k.A \Longrightarrow k= \frac{m}{A.\epsilon_{0}}}$$

Entonces, reemplazando por:
1. **Permitividad del vacío** $\epsilon_{0}$: ${8,85.10^{-12} \frac{F}{m}}$
2. **Área de las placas metálicas**: ${20cm x 20cm = 0,2m x 0,2m= 0,04m^{2}}$
3. **Pendiente**: ${1,30.10^{-12}}$
$${k= \frac{1,30.10^{-12}}{0,04m^{2}.8,85.10^{-12}}= 3,64406796}$$

Se calculó el **error de la constante dieléctrica _k_**: $${\Delta_{m}= \Delta_{k}.\epsilon_{0}.A \Longrightarrow \Delta_{k}= \frac{\Delta_{m}}{\epsilon_{0}.A} =  \frac{0,075.10^{-12}}{8,85.10^{-12}.0,04} = 0,21}$$

Redondeando el valor de _k_ a la misma cantidad de decimales que su error, nos queda como saldo final que **la constante dieléctrica _k_** es igual a $\Longrightarrow$ ${k = (3,6 \pm 0,2)}$. 
Utilizando la fórmula ${k = 1 + X_{e}}$ para calcular la susceptibilidad dieléctrica nos queda como resultado que: $${X_{e} = k - 1 = 3,6 - 1 = 2,6} \Longrightarrow {X_{e}= (2,6 \pm 0,2)}$$

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Valores de la nueva tabla (Acrílico)
ACRILICO_A = 1/0.0019
ACRILICO_D = 1/0.0037
ACRILICO_E = 1/0.0045
ACRILICO_ED = 1/0.0082
ACRILICO_EDA = 1/0.0101

CAPACIDAD_A = 4.55e-10
CAPACIDAD_D = 2.93e-10
CAPACIDAD_E = 2.3e-10
CAPACIDAD_ED = 1.47e-10
CAPACIDAD_EDA = 1.24e-10

d_values = [ACRILICO_D, ACRILICO_A, ACRILICO_E, ACRILICO_ED, ACRILICO_EDA]  # Valores de d [m]
C_values = [CAPACIDAD_D, CAPACIDAD_A, CAPACIDAD_E, CAPACIDAD_ED, CAPACIDAD_EDA]  # Valores de C [F]

# Ajuste lineal (recta) a los datos
coef = np.polyfit(d_values, C_values, 1)
poly1d_fn = np.poly1d(coef)  # Función polinómica de ajuste

# Crear el gráfico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='yellow', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la línea de ajuste
plt.plot(d_values, poly1d_fn(d_values), '--', color='blue', label="Ajuste lineal")

# Configuraciones del gráfico
plt.title('Acrílico')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 5.0e-10)
plt.legend()
plt.grid(True)

# Formato científico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje y

# Ajustar márgenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuación de la recta y R² fuera del gráfico
m, b = coef
plt.text(1.5, 4.5e-10, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='blue', ha='right')
R2 = np.corrcoef(d_values, C_values)[0, 1]**2
plt.text(1.5, 4.3e-10, f'$R^2$ = {R2:.2f}', fontsize=14, color='blue', ha='right')

# Mostrar el gráfico
plt.show()
```

1. **Pendiente** = ${7,74.10^{-13}}$
2. **Valor del coeficiente de determinación** ${R^{2}}$ = ${0,99}$
3. **Número de mediciones _N_** = ${5}$
4. **Permitividad del vacío** $\epsilon_{0}$ = ${8,85.10^{-12} \frac{F}{m}}$
5. **Área de las placas metálicas**: ${20cm x 20cm = 0,2m x 0,2m= 0,04m^{2}}$

Se calcula el **error de la pendiente $\Delta_{m}$**: $${\Delta_{m} = \sqrt{\frac{m^{2}}{(N -2)}.(\frac{1}{R^{2}} - 1)}}  \Longrightarrow \sqrt{\frac{(7,74.10^{-13})^{2}}{(5 -2)}.(\frac{1}{0,99} - 1)} = 4,49.10^{-14} = 0,45.10^{-13}$$

La **pendiente + su error** quedan definidos de la siguiente manera: ${m = (7,74.10^{-13} \pm 0,45.10^{-13})}$

Utilizando la fórmula de la pendiente de la recta para despejar la **constante dieléctrica _k_** se llegó a que $${m = \epsilon_{0}.k.A \Longrightarrow k= \frac{m}{A.\epsilon_{0}} = \frac{7,74.10^{-13}}{0,04m^{2}.8,85.10^{-12}} = 2,186440678}$$

Se calculó el **error de la constante dieléctrica _k_**: $${\Delta_{m}= \Delta_{k}.\epsilon_{0}.A \Longrightarrow \Delta_{k}= \frac{\Delta_{m}}{\epsilon_{0}.A} =  \frac{0,45.10^{-13}}{8,85.10^{-12}.0,04} = 0,1271186440}$$

Redondeando el valor de _k_ a la misma cantidad de decimales que su error, nos queda como saldo final que **la constante dieléctrica _k_** es igual a $\Longrightarrow$ ${k = (2,2 \pm 0,1)}$. 
Utilizando la fórmula ${k = 1 + X_{e}}$ para calcular la susceptibilidad dieléctrica nos queda como resultado que: $${X_{e} = k - 1 = 2,2 - 1 = 1,2} \Longrightarrow {X_{e}= (1,2 \pm 0,1)}$$

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# Valores de la tabla (Tercera configuración: acrílico + aire)
CUADRADOS_F = 1/0.01
CUADRADOS_FC = 1/0.0129
CUADRADOS_FE = 1/0.0145
CUADRADOS_FB = 1/0.01235
CUADRADOS_EC = 1/0.0074

CAPACIDAD_F = 4.87e-11
CAPACIDAD_FC = 3.98e-11
CAPACIDAD_FE = 3.66e-11
CAPACIDAD_FB = 4.09e-11
CAPACIDAD_EC = 6.14e-11

d_values = [CUADRADOS_F, CUADRADOS_FC, CUADRADOS_FE, CUADRADOS_FB, CUADRADOS_EC]  # Valores de 1/d [1/m]
C_values = [CAPACIDAD_F, CAPACIDAD_FC, CAPACIDAD_FE, CAPACIDAD_FB, CAPACIDAD_EC]  # Valores de C [F]

# Ajuste lineal (recta) a los datos
coef = np.polyfit(d_values, C_values, 1)
poly1d_fn = np.poly1d(coef)  # Función polinómica de ajuste

# Crear el gráfico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='green', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la línea de ajuste
plt.plot(d_values, poly1d_fn(d_values), '--', color='darkgreen', label="Ajuste lineal")

# Configuraciones del gráfico
plt.title('Acrílico + Aire')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 8.0e-11)
plt.legend()
plt.grid(True)

# Formato científico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato científico para el eje y

# Ajustar márgenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuación de la recta y R² fuera del gráfico
m, b = coef
plt.text(115, 7.5e-11, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='darkgreen', ha='right')
R2 = np.corrcoef(d_values, C_values)[0, 1]**2
plt.text(115, 7.1e-11, f'$R^2$ = {R2:.2f}', fontsize=14, color='darkgreen', ha='right')

# Mostrar el gráfico
plt.show()
```

Redondeando el valor de _k_ a la misma cantidad de decimales que su error, nos queda como saldo final que **la constante dieléctrica _k_** es igual a $\Longrightarrow$ ${k = (1,07 \pm 0,02)}$. 
Utilizando la fórmula ${k = 1 + X_{e}}$ para calcular la susceptibilidad dieléctrica nos queda como resultado que: $${X_{e} = k - 1 = 1,07 - 1 = 0,07} \Longrightarrow {X_{e}= (0,07 \pm 0,02)}$$

Este valor es muy cercano al del aire (${k_{aire} \approx 1,0006}$, ${X_{e} \approx 0}$), lo cual es esperable ya que el espacio entre las placas está ocupado casi en su totalidad por aire.

## Conclusiones
Tras el experimento se pudo concluir que la capacitancia no depende de la carga ni del voltaje aplicado, ya que ambos varían de forma proporcional y su cociente se mantiene constante. La capacitancia depende únicamente de la **geometría** del capacitor (el área de las placas y la distancia que las separa) y de la **permitividad del medio** entre ellas (dieléctrico o vacío).

En las tres configuraciones, los gráficos de ${C}$ en función de ${\frac{1}{d}}$ dieron un comportamiento lineal, con coeficientes de determinación ${R^{2} \geq 0,99}$. Esto confirma experimentalmente la relación ${C = \frac{\epsilon_{0}.k.A}{d}}$: **la capacidad es inversamente proporcional a la separación entre las placas**. Al aumentar el espesor del dieléctrico (por ejemplo, al apilar dos placas), la capacidad disminuye.

A partir de la pendiente de cada recta se obtuvieron los siguientes resultados:

Configuración | Constante dieléctrica _k_ | Susceptibilidad ${X_{e}}$
------------ | ------------ | -------------
Vidrio | ${(3,6 \pm 0,2)}$ | ${(2,6 \pm 0,2)}$
Acrílico | ${(2,2 \pm 0,1)}$ | ${(1,2 \pm 0,1)}$
Acrílico + aire | ${(1,07 \pm 0,02)}$ | ${(0,07 \pm 0,02)}$

* El **vidrio** es el material con mayor constante dieléctrica. Es decir, es el que más se polariza ante el campo eléctrico y, para una misma separación, permite almacenar más carga.
* El **acrílico** tiene una constante dieléctrica menor que la del vidrio, pero mayor que la del aire.
* En la **tercera configuración**, el espacio entre las placas está ocupado casi en su totalidad por aire. Se obtuvo ${k \approx 1}$, un valor muy cercano al del vacío (${k_{aire} \approx 1,0006}$). El pequeño exceso puede atribuirse a los cuadrados de acrílico de las esquinas. Este resultado sirve además como control del método: sin un dieléctrico sólido, el valor obtenido coincide con el esperado teóricamente.

Si se comparan con los valores de tablas (${k_{vidrio} \approx 4 - 7}$ y ${k_{acrílico} \approx 2,6 - 3,5}$), los valores obtenidos resultan algo menores. Las posibles fuentes de error son:
* **Capas de aire entre las placas**: las placas metálicas y las de vidrio o acrílico no son perfectamente planas, por lo que quedan finas capas de aire entre ellas. Estas capas (con ${k \approx 1}$) quedan en serie con el dieléctrico y reducen la capacidad efectiva. El efecto se acentúa al apilar varias placas. Por eso se usaron pesas para presionar el conjunto.
* **Capacidades parásitas y efectos de borde**: según la teoría, la ordenada al origen de los ajustes debería ser nula; sin embargo, se obtuvieron valores del orden de ${10^{-11} - 10^{-10}}$ F. Esto indica una capacidad adicional aportada por los cables y el multímetro, y por el campo que se "escapa" por los bordes de las placas (el modelo supone un campo uniforme y placas infinitas).
* **Errores de medición**: incertidumbre en la medición de los espesores con el calibre y la resolución propia del multímetro.

En síntesis, se cumplió el objetivo del trabajo práctico: se verificó la dependencia ${C \propto \frac{1}{d}}$, se determinó la susceptibilidad dieléctrica de cada material y se comprobó que el vidrio es mejor dieléctrico que el acrílico, y que ambos superan al aire.
