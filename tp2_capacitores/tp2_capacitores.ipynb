## 3. Introducción
Un capacitor es un dispositivo formado por dos placas o láminas conductoras separadas entre sí por un dieléctrico, especialmente diseñado para almacenar una cantidad finita de energía. La unidad de la capacidad es el Faradio ${[F]}$.

Un dieléctrico es un medio eléctricamente aislante que se opone al paso de la corriente eléctrica.

Si se conecta el capacitor a un generador se produce un pasaje de electrones hasta que las dos placas consiguen albergar el mismo número de cargas pero de _signo contrario_. Una vez cargado, la corriente que circula por el circuito es nula y, aunque se desconecte el generador, el capacitor no se descarga.

La forma más fácil de armar un capacitor es colocar dos placas paralelas enfrentadas entre sí, separadas por una distancia que es pequeña en comparación con sus dimensiones. Con una carga $q$ repartida de forma uniforme en cada placa de área $A$ (densidad superficial $\sigma=q/A$), y usando la Ley de Gauss o superposición de campos, el campo eléctrico ${E}$ entre las placas es uniforme y vale: $${E = \frac{\sigma}{\epsilon_{0}} = \frac{q}{\epsilon_{0}.A}}$$
donde ${\epsilon_{0}}$ es la **permitividad eléctrica en el vacío**, con un valor de ${8,85.10^{-12} \frac{F}{m}}$.

Dado que el campo es uniforme y la distancia entre las placas es _d_, la diferencia de potencial entre ambas será: $${\Delta V_{ab}= E . d = \frac{q.d}{\epsilon_{0}.A}}$$

Entonces, la **capacidad de un capacitor plano en el _vacío_** queda definida como: $${C= \frac{q}{\Delta V_{ab}} = \frac{\epsilon_{0}.A}{d}}$$

La capacitancia varía si, en vez de vacío, hay otro **tipo de _material_ entre las placas**: se ve afectada por las características de ese material, así que su ecuación cambia a: $${C= \frac{\epsilon_{0}.k.A}{d} \Longrightarrow C= \frac{\epsilon_{0}.(1+X_{e}).A}{d}}$$
Allí $k$ es la **constante dieléctrica** del material entre las placas, y $X_{e}=k-1$ su **susceptibilidad dieléctrica**.

Como $C$ depende de $1/d$ y no de $d$ directamente, conviene graficar $C$ en función de $1/d$ en vez de $d$: así los puntos se alinean en una **recta** ($C=(\epsilon_{0}.k.A)\cdot\frac{1}{d}$), y la pendiente de esa recta es directamente $\epsilon_{0}.k.A$ — de ahí se puede despejar $k$. Por eso todos los gráficos que siguen son de $C$ vs. $1/d$.

## 5. Mediciones y resultados

### 5.1. Primera configuración — Acrílico en las esquinas + aire

Se utilizaron dos placas metálicas con separaciones de 4 cuadrados de acrílico pequeños ubicados en cada esquina de las placas metálicas, dejando pasar aire entre las mismas. La medición se realizó en picoFaradios y luego se convirtió a Faradios.

   

Acrílico + aire | d [m] | C [F]
------------ | ------------ | -------------
Cuadrados E+C | 0,0074m| ${6,14.10^{-11}}$ F
Cuadrados F   | 0,01m| ${4,87.10^{-11}}$ F
Cuadrados F+B | 0,01235m | ${4,09.10^{-11}}$ F
Cuadrados F+C  | 0,0129m| ${3,98.10^{-11}}$ F
Cuadrados F+E | 0,0145m| ${3,66.10^{-11}}$ F

Con los datos obtenidos se realizó un gráfico con los valores de la capacidad _C [F]_ por los valores de $\frac{1}{d}$ (_d_ es la separación entre las placas metálicas), para así determinar la susceptibilidad dieléctrica $X_{e}$


```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker
from scipy import stats

# Valores de la tabla (Primera configuracion: acrilico + aire)
CUADRADOS_F = 1/0.01
CUADRADOS_FC = 1/0.0129
CUADRADOS_FE = 1/0.0145
CUADRADOS_FB = 1/0.01235
CUADRADOS_EC = 1/0.0074

CAPACIDAD_F = 4.87e-11
CAPACIDAD_FC = 3.98e-11
CAPACIDAD_FE = 3.66e-11
CAPACIDAD_FB = 4.09e-11
CAPACIDAD_EC = 6.14e-11

d_values = [CUADRADOS_F, CUADRADOS_FC, CUADRADOS_FE, CUADRADOS_FB, CUADRADOS_EC]  # Valores de 1/d [1/m]
C_values = [CAPACIDAD_F, CAPACIDAD_FC, CAPACIDAD_FE, CAPACIDAD_FB, CAPACIDAD_EC]  # Valores de C [F]

# Ajuste lineal (recta) por cuadrados minimos, con scipy: da la pendiente, la ordenada
# al origen, R y tambien el error estandar de la pendiente (stderr)
ajuste = stats.linregress(d_values, C_values)
m, b = ajuste.slope, ajuste.intercept
R2 = ajuste.rvalue ** 2
delta_m = ajuste.stderr

print(f"m (pendiente)        = {m:.4e}")
print(f"b (ordenada)         = {b:.4e}")
print(f"R2                   = {R2:.4f}")
print(f"Delta_m (stderr)     = {delta_m:.4e}")

# Crear el grafico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='green', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la linea de ajuste
x_recta = np.array(d_values)
plt.plot(x_recta, m * x_recta + b, '--', color='darkgreen', label="Ajuste lineal")

# Configuraciones del grafico
plt.title('Acrilico + Aire')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 8.0e-11)
plt.legend()
plt.grid(True)

# Formato cientifico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato cientifico para el eje y

# Ajustar margenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuacion de la recta y R2 en el grafico
plt.text(115, 7.5e-11, f'y = {m:.2e}x + {b:.2e}', fontsize=14, color='darkgreen', ha='right')
plt.text(115, 7.1e-11, f'$R^2$ = {R2:.2f}', fontsize=14, color='darkgreen', ha='right')

# Mostrar el grafico
plt.show()
```

Como vimos en la Introducción, la pendiente de este ajuste es $m=\epsilon_{0}.k.A$, y la calculamos arriba con `scipy.stats.linregress` (el mismo cálculo por cuadrados mínimos que antes hacíamos con `np.polyfit`, pero que de paso ya devuelve el error de la pendiente): $$m = \epsilon_{0}.k.A = 3,77.10^{-13}$$

Esa función calcula el error de la pendiente (`stderr`) con esta fórmula, en función del número de mediciones $N$ y de qué tan bien se acomodan los puntos a la recta ($R^{2}$, el coeficiente de determinación: va de 0 a 1, y cuanto más cerca de 1, menos dispersos están los puntos respecto de la recta):
$${\Delta_{m} = \sqrt{\frac{m^{2}}{N-2}\left(\frac{1}{R^{2}}-1\right)}}$$
Con $N=5$ y $R^{2}=0,9994$ (el ajuste es tan bueno que redondeado a dos decimales da $1,00$, por eso lo dejamos con más decimales): $${\Delta_{m} = \sqrt{\frac{(3,77.10^{-13})^{2}}{5-2}\left(\frac{1}{0,9994}-1\right)} = 5,28.10^{-15} = 0,05.10^{-13}}$$
(el mismo valor que imprime el código de arriba).

Entonces: $${m = (3,77 \pm 0,05).10^{-13}}$$

Despejando la **constante dieléctrica _k_** de $m=\epsilon_{0}.k.A$, con $\epsilon_{0}=8,85.10^{-12}\,F/m$ y $A=0,2m\times0,2m=0,04\,m^{2}$: $${k= \frac{m}{A.\epsilon_{0}} = \frac{3,77.10^{-13}}{0,04.8,85.10^{-12}} = 1,065}$$

El error de $k$ sale de propagar el de $m$ (ya que $A$ y $\epsilon_{0}$ son constantes): $${\Delta_{k}= \frac{\Delta_{m}}{\epsilon_{0}.A} =  \frac{5,28.10^{-15}}{8,85.10^{-12}.0,04} = 0,0149}$$

Redondeando el error a una cifra significativa (siempre hacia arriba) y $k$ a la misma cantidad de decimales, **la constante dieléctrica** es $k = (1,06 \pm 0,02)$, y como $X_{e}=k-1$: $${X_{e}= (0,06 \pm 0,02)}$$

Acá el separador no es un único material homogéneo entre las placas: son cuadraditos de acrílico solo en las esquinas, y el resto del área entre placas queda con aire. Por eso el valor de $k$ da tan cerca del aire ($k \approx 1$, $X_{e} \approx 0$): como el acrílico ocupa una fracción chica del área entre placas, el dieléctrico _efectivo_ de esta configuración está dominado por el aire (más adelante vamos a ver que el acrílico puro, sin aire de por medio, da un $k$ bastante más alto).

Repetimos el mismo análisis que en la configuración anterior: gráfico de $C$ vs. $\frac{1}{d}$ y ajuste lineal.

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker
from scipy import stats

# Valores de la tabla
VIDRIO_B = 1/0.0038
VIDRIO_C = 1/0.0059
VIDRIO_BB = 1/0.0076
VIDRIO_BC = 1/0.0097
VIDRIO_CC = 1/0.0118

CAPACIDAD_B = 4.3e-10
CAPACIDAD_C = 3.26e-10
CAPACIDAD_BB = 2.47e-10
CAPACIDAD_BC = 2.3e-10
CAPACIDAD_CC = 2.02e-10

d_values = [VIDRIO_C, VIDRIO_B, VIDRIO_BC, VIDRIO_CC, VIDRIO_BB]  # Valores de 1/d [m]
C_values = [CAPACIDAD_C, CAPACIDAD_B, CAPACIDAD_BC, CAPACIDAD_CC, CAPACIDAD_BB]  # Valores de C [F]

# Ajuste lineal (recta) por cuadrados minimos, con scipy: da la pendiente, la ordenada
# al origen, R y tambien el error estandar de la pendiente (stderr)
ajuste = stats.linregress(d_values, C_values)
m, b = ajuste.slope, ajuste.intercept
R2 = ajuste.rvalue ** 2
delta_m = ajuste.stderr

print(f"m (pendiente)        = {m:.4e}")
print(f"b (ordenada)         = {b:.4e}")
print(f"R2                   = {R2:.4f}")
print(f"Delta_m (stderr)     = {delta_m:.4e}")

# Crear el grafico
plt.figure(figsize=(10, 6))  # Aumentar el ancho de la figura
plt.scatter(d_values, C_values, color='orange', label="Datos experimentales", s=100, edgecolor='black')

# Agregar la linea de ajuste
x_recta = np.array(d_values)
plt.plot(x_recta, m * x_recta + b, '--', color='purple', label="Ajuste lineal")

# Configuraciones del grafico
plt.title('Vidrio')
plt.xlabel('1/d [1/m]')
plt.ylabel('C [F]')
plt.ylim(0, 5.0e-10)
plt.legend()
plt.grid(True)

# Formato cientifico para los ejes
ax = plt.gca()
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato cientifico para el eje y
ax.xaxis.set_major_formatter(ticker.FormatStrFormatter('%.1e'))  # Formato cientifico para el eje x

# Ajustar margenes
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)

# Mostrar la ecuacion de la recta y R2 (arriba a la izquierda, con el hueco que deja el dato)
ax.text(0.03, 0.95, f'y = {m:.2e}x + {b:.2e}', transform=ax.transAxes, fontsize=14, color='purple', ha='left', va='top')
ax.text(0.03, 0.88, f'$R^2$ = {R2:.2f}', transform=ax.transAxes, fontsize=14, color='purple', ha='left', va='top')

# Mostrar el grafico
plt.show()
```

Repetimos el mismo método que en la configuración anterior (`scipy.stats.linregress`). Del ajuste: $m=(1,30\pm0,089).10^{-12}$ (con $N=5$, $R^{2}=0,9860$).

Despejando $k$ y propagando el error igual que antes: $${k= \frac{m}{A.\epsilon_{0}} = \frac{1,30.10^{-12}}{0,04.8,85.10^{-12}} = 3,66 \qquad \Delta_{k}=\frac{\Delta_{m}}{\epsilon_{0}.A}=0,25}$$

Redondeando $k$ a la misma cantidad de decimales que su error: $k = (3,7 \pm 0,3)$, y $${X_{e}= (2,7 \pm 0,3)}$$

## 6. Conclusiones

Tras el experimento se pudo concluir que la capacitancia no depende de la carga ni del voltaje aplicado, ya que ambos varían proporcionalmente entre sí y su cociente ($C = q/\Delta V$) se mantiene constante. En cambio, la capacitancia sí depende de factores geométricos —el área de las placas y la distancia que las separa— y del medio dieléctrico ubicado entre ellas, tal como predice ${C = \frac{\epsilon_{0}.k.A}{d}}$.

Esto se vio reflejado directamente en los resultados: en las tres configuraciones los puntos de $C$ vs. $1/d$ se alinearon en una recta (la capacidad es inversamente proporcional a la separación entre placas), y cada configuración dio una pendiente distinta, y por lo tanto una $k$ y una $X_{e}$ distintas.

| Configuración | Separador | $k$ | $X_{e}$ |
|---|---|---:|---:|
| 1 | Acrílico (esquinas) + aire | $1,06 \pm 0,02$ | $0,06 \pm 0,02$ |
| 2 | Vidrio | $3,7 \pm 0,3$ | $2,7 \pm 0,3$ |
| 3 | Acrílico | $2,2 \pm 0,1$ | $1,2 \pm 0,1$ |

El vidrio resultó ser el dieléctrico con mayor $X_{e}$ de los tres, seguido por el acrílico. La primera configuración, al tener acrílico solo en una fracción chica del área entre las placas (con aire en el resto), dio una $X_{e}$ mucho menor, cercana a la del aire ($k_{aire} \approx 1,0006$). Esto confirma que la capacitancia depende del material dieléctrico usado, y que lo que importa específicamente no es solo qué materiales están presentes entre las placas sino cómo se distribuyen espacialmente en el área que ocupan.

Si se comparan con valores de tabla ($k_{vidrio} \approx 4 - 7$ y $k_{acrílico} \approx 2,6 - 3,5$), los valores obtenidos para el vidrio y el acrílico resultan algo menores. Como fuentes de error se pueden mencionar:
* El **contacto imperfecto** entre las placas metálicas y los separadores, que puede dejar pequeños huecos de aire y aumentar la $d$ efectiva por encima de la medida con el calibre. Esas capas de aire ($k \approx 1$) quedan en serie con el dieléctrico y bajan la capacidad medida, lo que explica que $k$ dé por debajo del valor de tabla. El efecto es mayor cuando se apilan varias láminas.
* **Capacidades parásitas**: según el modelo, la ordenada al origen de los ajustes debería ser nula, pero dio del orden de $10^{-11} - 10^{-10}\,F$. Esto indica una capacidad adicional, aportada por los cables y el propio capacímetro, que no depende de $d$.
* Las placas metálicas no son infinitas, por lo que existen **efectos de borde** no contemplados por el modelo ideal.
* La **precisión** del capacímetro digital y del calibre.

Aun así, los ${R^{2}}$ obtenidos en los tres ajustes (entre $0,986$ y $0,999$) muestran que el modelo lineal ${C \propto 1/d}$ describe bien el comportamiento medido en las tres configuraciones.
